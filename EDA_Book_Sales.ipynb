## EDA of Book Sales Data

In this project, we are going to perform EDA on Book Sales Data & Answer the following Questions
1) Show the trend of number of units sold over Publishing year
2) Show the relation between sale price & units sold
3) Show the relation between Average book rating & units sold
4) Show the relation between Average book rating & Book ratings count
5) Show the distribution of units sold genre wise
6) Show the composition of no of books genrewise
7) Who are the top 5 authors as per Avg book rating
8) Who are the bottom 5 authors as per Avg book rating
9) Which are the top 10 Publishers with highest revenue
10) Show the distribution of units sold genre wise & Author rating wise



```python
## Importing neccessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
## Loading the CSV file
df = pd.read_csv("Books_Data_Clean.csv")
```

```python
df.head()
```

```python
df.describe() 
'''there are negetive entries in Publishing Year column 
& the values are in float format which has to corrected'''
```

```python
df = df[df["Publishing Year"]>1900] ## considering books with publishing year greater than 1900
```

```python
df["Publishing Year"].astype(int) ##coverting float datatype to int
```

```python
## Checking for null values & deleting it if present
df.isna().sum().sum()
df.dropna()
```

```python
## Checking for duplicate values & deleting it if present
df.duplicated().sum() ## There are no duplicate values
```

1) Show the trend of number of units sold over Publishing year

```python
units_by_year = df.groupby("Publishing Year")["units sold"].sum().reset_index()
units_by_year
```

```python
sns.lineplot(x="Publishing Year",y="units sold",data=units_by_year)
plt.title("Trend of units sold over years")

'''Most of the units are sold (around 59%) after the year 2000'''
```

2) Show the relation between sale price & units sold

```python
sns.scatterplot(x="sale price",y="units sold",data=df)
plt.title("Relation between sale price & units sold")
'''There is no significant relation between units sold & sale price. 
Some of the books with lesser price have less than 10000 units sold 
while other books with same price have more than 30000 units sold'''
```

3) Show the relation between Average book rating & units sold

```python
sns.scatterplot(x="Book_average_rating",y="units sold",data=df)
plt.title("Relation between Avg book ratings & units sold")
'''There is no significant relation between units sold & Avg book ratings. 
Some of the books with lesser Avg Rating have less than 10000 units sold 
while other books with Avg Rating have more than 30000 units sold'''
```

4) Show the relation between Average book rating & Book ratings count


```python
sns.scatterplot(x="Book_average_rating",y="Book_ratings_count",data=df)
plt.title("Relation between Avg book ratings & Book ratings count")
'''There is a direct relation between Avg book ratings & Book ratings count.
As the number of ratings increases, the Avg book rating increases'''
```

5) Show the distribution of units sold genre wise

```python
sns.boxplot(x="genre",y="units sold",data=df) ## Box plot with outliers
```

```python
sns.boxplot(x="genre",y="units sold",data=df,showfliers=False) ## Box plot without outliers
'''Among all genres, Children books have the most widespread units sold Data.
ranging from 106 units sold to 50112 units sold'''
```

6) Show the composition of no of books genrewise

```python
Genre_wise_books=df.groupby("genre")["index"].count().reset_index()
Genre_wise_books.columns = ["Genre","No of books"]
Genre_wise_books
```

```python
# Pie chart for Genre wise books
plt.pie(
    Genre_wise_books["No of books"],   
    labels=Genre_wise_books["Genre"],  
    autopct='%1.1f%%',                 
    startangle=90,                     
    shadow=True,                       
    explode=[0.05]*len(Genre_wise_books) 
)

plt.title("Genre-wise Book Distribution")
plt.show()
''' around 77% of the books are of genre fiction genre'''
```

7) Who are the top 5 authors as per Avg book rating

```python
Top_5_Authors = df.groupby("Author")["Book_average_rating"].mean().reset_index()\
.sort_values("Book_average_rating",ascending=False).head()

'''Bill Watterson is the best Author with Avg rating of 4.65'''

Top_5_Authors
```

8) Who are the bottom 5 authors as per Avg book rating

```python
Bottom_5_Authors = df.groupby("Author")["Book_average_rating"].mean().reset_index()\
.sort_values("Book_average_rating",ascending=True).head()

'''Sue Monk Kidd is the Worst Author with Avg rating of 3.10'''

Bottom_5_Authors
```

9) Which are the top 10 Publishers with highest revenue

```python
Top_10_Publishers = df.groupby("Publisher ")["publisher revenue"].sum()\
.reset_index().sort_values("publisher revenue",ascending=False).head(10)
sns.barplot(x="publisher revenue",y="Publisher ",data=Top_10_Publishers,color="#248ae9")
plt.title("Top 10 Publishers by Revenue")
```

10) Show the distribution of units sold genre wise & Author rating wise

```python
pvt1=df.pivot_table(values="units sold",index="genre",columns="Author_Rating",aggfunc="sum")
plt.figure(figsize=(10,6))
sns.heatmap(
    pvt1, 
    annot=True,        
    fmt=".0f",         
    cmap="Blues",      
    linewidths=0.5,    
    cbar_kws={'label': 'Units Sold'}  
)

plt.title("Units Sold by Genre vs Author Rating")
plt.xlabel("Author Rating")
plt.ylabel("Genre")
plt.show()
'''In genre fiction more units are sold with Excellent & intermediate author rating
There are very less ubits sold with Famous & Novice Author rating across genres'''
```
